```python
import requests
response = requests.get("https://api.github.com/")
print(response.status_code)
```

```python

# Making a quick check to see if the request was successful
if response.status_code == 200:
    print("Success")
elif response.status_code == 404:
    print("Not found")
```

```python
# Searching github for all repos with language == python and requests in readme.md or title
response = requests.get(
    "https://api.github.com/search/repositories",
    params = {"q":"requests+language:python"}
)
response.json()
```

```python
# Getting an image
response = requests.get('https://imgs.xkcd.com/comics/making_progress.png')
response.content # is in bytes

# Then showing the img
import os
import matplotlib.pyplot as plt
with open(r'img.png','wb') as f:
    f.write(response.content)

img = plt.imread("img.png")
plt.imshow(img , cmap = "gray")
```
